```python
#pre filtering
#now we will remove the users that has less rated movies
#we will remove the movies that has been rated less in numbers
```

```python
#dataframe of movie_summary have all movie id and its counts
```

```python
movie_summary = netflix_dataset.groupby("MovieID")["Rating"].agg(["count"])
```

```python
movie_summary
```

```python
#we will set up benchmark=>60 percentile of counts
```

```python
movie_benchmark = round(movie_summary['count'].quantile(0.6),0)
movie_benchmark
```

```python
#drop_movie_list => keep all the movies less than count of benchmark
```

```python
drop_movie_list = movie_summary[movie_summary['count']<movie_benchmark].index
```

```python
drop_movie_list
```

```python
len(drop_movie_list)
```

```python
4499-2699
```

```python
#apply the same operation for customer as well, and find drop_cust_list
```

```python
cust_summary = netflix_dataset.groupby("CustID")["Rating"].agg(["count"])
```

```python
cust_summary
```

```python
cust_benchmark = round(cust_summary['count'].quantile(0.6),0)
cust_benchmark
```

```python
drop_cust_list = cust_summary[cust_summary['count']<cust_benchmark].index
```

```python
drop_cust_list
```

```python
len(drop_cust_list)
```

```python
#from netflix_dataset remove dropcustlist and dropmovielist
```

```python
print("The shape of data before triming:",netflix_dataset.shape)
netflix_dataset = netflix_dataset[~netflix_dataset['MovieID'].isin(drop_movie_list)]
netflix_dataset = netflix_dataset[~netflix_dataset['CustID'].isin(drop_cust_list)]
print("The shape of data after triming:",netflix_dataset.shape)
```
