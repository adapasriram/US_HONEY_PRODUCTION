Is the average purchase made by men of age 18-25 is 10000 .still it is same?

```python
men_18_25 = df[(df['Age']==1) & (df['Gender']==1)]
men_18_25
```

```python
sample_size = 1100
```

```python
men_18_25_sample = men_18_25.sample(sample_size, random_state=0)
men_18_25_sample
```

```python
from statsmodels.stats.weightstats import ztest

z_stat,p_value = ztest(men_18_25_sample['Purchase'],value=10000)
```

```python
p_value
```

```python
if p_value<0.05:
    print("We have to reject the null hypothesis, Average purchase made by men of Age group 18-25 is not 10000")
else:
    print("We failed to reject the null hypothesis,the average purchase made by men of Age group 18-25 is 10000.")
```

Is the average purchase made by men and women of age 18-25 same.

```python
from scipy.stats import ttest_ind
```

```python
data_men = df.loc[(df['Gender']==1) & (df['Age']==1)] #Selecting all male of 18-25 age category
data_women = df.loc[(df['Gender']==0) & (df['Age']==1)] #Selectin all females of 18-25 age category
```

```python

data_men.shape
```

```python
data_women.shape
```

```python
# data_men_sample = 1100
# data_women_sample = 340
# A) 1100 B) 340 C) Average It's better to choose  a value between 1100 and 340
sample_size = 700
data_men_sample = data_men.sample(sample_size,random_state=34)
data_women_sample = data_women.sample(sample_size,random_state=34)
```

```python
data_men_sample
```

```python
data_women_sample
```

```python
t_stat, p_value = ttest_ind(data_men_sample['Purchase'],data_women_sample['Purchase'],equal_var=False)


#equal_var = False
# Welch's t-test
```

```python
p_value
```

```python
if p_value<0.05:
    print("We have to reject the null hypothesis, Average purchase made by both the groups is not same.")
else:
    print("We failed to reject the null hypothesis, Average purchase made by both the groups is same.")
```
